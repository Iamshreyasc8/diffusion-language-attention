```python
import pandas as pd
from transformers import pipeline

# Step 1: Load the Excel file
file_path = '/content/drive/MyDrive/Thesis/dataset/support_verb/filtered_support_verb.xlsx'  # Replace with the uploaded file name
df = pd.read_excel(file_path)

# Step 2: Load a Hugging Face model for text generation
# Using T5 for summarization and rephrasing
rephrase_model = pipeline("text2text-generation", model="t5-small")

# Step 3: Define a function to enhance prompts
def enhance_prompt(Context, Expression):
    try:
        # Prepare the input for the model
        input_text = (
            f"Rephrase and enhance this prompt to make it concise, meaningful, and creative, "
            f"while focusing on the phrase '{Expression}': {Context}"
        )

        # Generate the response
        result = rephrase_model(input_text, max_length=50, num_return_sequences=1, truncation=True)
        return result[0]['generated_text']
    except Exception as e:
        print(f"Error processing prompt: {e}")
        return "Error processing prompt"

# Step 4: Apply the function to the DataFrame
df['ai_sorted'] = df.apply(
    lambda row: enhance_prompt(row['Context'], row['Expression']), axis=1
)

# Step 5: Save the updated DataFrame
output_file_path = '/content/drive/MyDrive/Thesis/dataset/support_verb/ai-filtered_support_verb.xlsx'
df.to_excel(output_file_path, index=False)

# Step 6: Notify the user
print(f"Processed file saved at: {output_file_path}")
```

```python
import pandas as pd
import re
from transformers import BartForConditionalGeneration, BartTokenizer

# Step 3: Load the BART model and tokenizer
tokenizer = BartTokenizer.from_pretrained('facebook/bart-large-cnn')
model = BartForConditionalGeneration.from_pretrained('facebook/bart-large-cnn')

# Step 4: Load the Excel file
file_path = '/content/drive/MyDrive/Thesis/dataset/support_verb/filtered_support_verb.xlsx'  # Replace with your actual file name
df = pd.read_excel(file_path)

# Step 5: Ensure column names are clean
df.columns = df.columns.str.strip()  # Remove leading/trailing spaces from column names

# Verify column names
print("Column Names:", df.columns)

# Check if the required columns exist
if 'Context' not in df.columns or 'Expression' not in df.columns:
    raise KeyError("The required columns 'Context' or 'Expression' are missing. Please check the file.")

# Step 6: Clean Context by removing prefixes (e.g., #42-64) and irrelevant details
def clean_context(context):
    # Remove leading # and number patterns like #42-64
    cleaned_context = re.sub(r'^#\d+-\d+\s*', '', context)
    # Remove verbose or irrelevant details
    cleaned_context = re.sub(r'\s*[.,]*\s*says\s.*|[.,]*according to\s.*', '', cleaned_context)
    return cleaned_context.strip()

# Step 7: Define a function to enhance prompts with an image-generation perspective
def enhance_prompt(context, expression):
    try:
        # Clean the context before processing
        cleaned_context = clean_context(context)

        # Combine cleaned context and expression into a single input
        input_text = (
            f"Simplify this for an image-generation model. Use descriptive, visual language and focus on the key idea '{expression}'. "
            f"Ensure the text is concise, coherent, and suitable for generating an image: {cleaned_context}"
        )

        # Tokenize the input text
        inputs = tokenizer.encode("summarize: " + input_text, return_tensors='pt', max_length=1024, truncation=True)

        # Generate the simplified, image-focused prompt
        summary_ids = model.generate(
            inputs,
            max_length=50,  # Limit length to ensure concise prompts
            min_length=15,  # Ensure a minimum length for meaningful prompts
            length_penalty=1.2,  # Slight penalty for longer outputs
            num_beams=4,  # Use beam search for diversity
            early_stopping=True  # Stop when the model is confident
        )

        # Decode the generated summary
        simplified_prompt = tokenizer.decode(summary_ids[0], skip_special_tokens=True)

        # Final cleanup to remove any residual abstract details
        simplified_prompt = re.sub(r'\s*[.,]*\s*says\s.*|[.,]*according to\s.*', '', simplified_prompt)

        # Return the image-focused prompt
        return simplified_prompt.strip()
    except Exception as e:
        print(f"Error processing prompt: {e}")
        return "Error processing prompt"

# Step 8: Apply the function to each row and create a new column 'ai_sorted'
df['Context'] = df['Context'].astype(str)  # Ensure 'Context' is treated as a string
df['Expression'] = df['Expression'].astype(str)  # Ensure 'Expression' is treated as a string

# Apply the cleaning and enhancement functions
df['ai_sorted'] = df.apply(lambda row: enhance_prompt(row['Context'], row['Expression']), axis=1)

# Step 9: Save the updated DataFrame to a new Excel file
output_file_path = '/content/ai_sorted_prompts_for_images.xlsx'
df.to_excel(output_file_path, index=False)

# Step 10: Notify the user and provide a download link
print(f"Processed file saved at: {output_file_path}")

```

```python
import pandas as pd
from transformers import T5Tokenizer, T5ForConditionalGeneration

# Load the flan-t5-large model and tokenizer
tokenizer = T5Tokenizer.from_pretrained("google/flan-t5-large")
model = T5ForConditionalGeneration.from_pretrained("google/flan-t5-large", device_map="auto")

def clean_context(context):
    """
    Removes unnecessary prefixes (e.g., #42-64) and irrelevant details.
    """
    import re
    # Remove prefixes like #42-64
    cleaned_context = re.sub(r'^#\d+-\d+\s*', '', context)
    return cleaned_context.strip()

def generate_image_prompt(context, expression):
    """
    Generates refined, concise, and visually descriptive prompts for image generation.
    """
    try:
        # Clean the context
        cleaned_context = clean_context(context)

        # Check for empty or missing values
        if not cleaned_context or pd.isna(cleaned_context):
            return "Error: Context is missing."
        if not expression or pd.isna(expression):
            return "Error: Expression is missing."

        # Construct refined instructions for the model with examples
        input_text = (
            f"Create a visually descriptive and concise image-generation prompt. Simplify the text, focusing on imagery and the expression: '{expression}'.\n\n"
            f"Example 1: Context: 'The Taliban, overthrown by U.S.-backed Afghan forces, were quick to take credit.' Expression: 'take credit'.\n"
            f"Output: 'The Taliban claimed responsibility for the attack after being overthrown by U.S.-backed forces.'\n\n"
            f"Example 2: Context: 'This attack was to make clear our reaction to Obama's trip to Afghanistan.' Expression: 'reaction'.\n"
            f"Output: 'An attack symbolizing a reaction to Obama's visit to Afghanistan.'\n\n"
            f"Now create a prompt for:\nContext: {cleaned_context}\nExpression: {expression}"
        )

        # Tokenize and prepare inputs
        input_ids = tokenizer(input_text, return_tensors="pt").input_ids.to("cuda")

        # Generate the output
        outputs = model.generate(
            input_ids,
            max_length=70,  # Limit the output length
            min_length=10,  # Ensure meaningful output
            num_beams=4,  # Use beam search for diverse outputs
            early_stopping=True  # Stop when confident
        )

        # Decode the output
        generated_prompt = tokenizer.decode(outputs[0], skip_special_tokens=True)

        return generated_prompt.strip()
    except Exception as e:
        print(f"Error generating prompt: {e}")
        return "Error generating prompt"

# Load the dataset
file_path = '/content/drive/MyDrive/Thesis/dataset/support_verb/filtered_support_verb.xlsx'  # Replace with your uploaded file path
df = pd.read_excel(file_path)

# Apply the refined prompt generation function to each row
df['ai_sorted'] = df.apply(
    lambda row: generate_image_prompt(row['Context'], row['Expression']),
    axis=1
)

# Save the processed dataset
output_file_path = '/content/drive/MyDrive/Thesis/dataset/support_verb/aafiltered_support_verb.xlsx'
df.to_excel(output_file_path, index=False)


print(f"Processed file saved at: {output_file_path}")
```

```python
import pandas as pd
from transformers import T5Tokenizer, T5ForConditionalGeneration
from tqdm import tqdm  # For progress tracking

# Step 1: Load the flan-t5-large Model and Tokenizer
tokenizer = T5Tokenizer.from_pretrained("google/flan-t5-large")
model = T5ForConditionalGeneration.from_pretrained("google/flan-t5-large", device_map="auto")

# Step 2: Define the Context Cleaning and Prompt Generation Functions
def clean_context(context):
    """
    Removes unnecessary prefixes (e.g., #42-64) from the context.
    """
    import re
    cleaned_context = re.sub(r'^#\d+-\d+\s*', '', context)
    return cleaned_context.strip()

def generate_image_prompt(context, expression, index, total):
    """
    Generates a concise and visually descriptive prompt for image generation.
    """
    try:
        # Clean the context
        cleaned_context = clean_context(context)

        # Check for empty or missing values
        if not cleaned_context or pd.isna(cleaned_context):
            return "Error: Context is missing."
        if not expression or pd.isna(expression):
            return "Error: Expression is missing."

        # Construct the input text for flan-t5-large
        input_text = f"paraphrase: {cleaned_context}. Focus on: {expression}."

        # Tokenize and prepare inputs
        input_ids = tokenizer(input_text, return_tensors="pt").input_ids.to("cuda")

        # Generate the output
        outputs = model.generate(
            input_ids,
            max_length=75,
            min_length=10,
            num_beams=4,
            early_stopping=True
        )

        # Decode the output
        generated_prompt = tokenizer.decode(outputs[0], skip_special_tokens=True)

        # Show progress
        print(f"Processed {index+1}/{total} rows", end="\r")

        return generated_prompt.strip()
    except Exception as e:
        print(f"Error at row {index+1}: {e}")
        return "Error generating prompt"

# Step 3: Load the Dataset
file_path = '/content/drive/MyDrive/Thesis/dataset/not_mwe/not_mwe.xlsx'
df = pd.read_excel(file_path)

# Step 4: Apply the Prompt Generation Function with Progress Tracking
total_rows = len(df)
tqdm.pandas()  # Enable progress bar

df['ai_sorted'] = df.progress_apply(
    lambda row: generate_image_prompt(row['Context'], row['Expression'], row.name, total_rows),
    axis=1
)

# Step 5: Save the Processed Dataset
output_file_path = '/content/drive/MyDrive/Thesis/dataset/not_mwe/sorted-not_mwe.xlsx'
df.to_excel(output_file_path, index=False)

print(f"\nProcessed file saved at: {output_file_path}")
```
